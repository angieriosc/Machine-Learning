```python
import pandas as pd
import joblib
from tensorflow.keras.models import load_model
```

```python
ruta_carpeta = "/content/drive/MyDrive/salary_checkpoints"

ruta_rf = ruta_carpeta + "/random_forest_salary_model.pkl"
ruta_nn = ruta_carpeta + "/best_salary_model.keras"
ruta_columnas = ruta_carpeta + "/columnas_modelo.pkl"

modelo_rf = joblib.load(ruta_rf)
modelo2 = load_model(ruta_nn)
columnas_modelo = joblib.load(ruta_columnas)

print("Modelos cargados correctamente")
```

```python
cgpa = float(input("Ingrese el CGPA: "))
college_tier = int(input("Ingrese el college tier: "))

python_skill = int(input("¿Skill Python? 1 = Sí, 0 = No: "))
dsa_skill = int(input("¿Skill DSA? 1 = Sí, 0 = No: "))
ml_skill = int(input("¿Skill Machine Learning? 1 = Sí, 0 = No: "))
web_dev_skill = int(input("¿Skill Web Development? 1 = Sí, 0 = No: "))

coding_score = float(input("Ingrese el coding score: "))
communication_score = float(input("Ingrese el communication score: "))
aptitude_score = float(input("Ingrese el aptitude score: "))

internships = int(input("Ingrese el número de internships: "))
projects = int(input("Ingrese el número de proyectos: "))
backlogs = int(input("Ingrese el número de backlogs: "))

resume_score = float(input("Ingrese el resume score: "))
skill_score = float(input("Ingrese el skill score: "))

company_type = input("Ingrese el tipo de compañía, por ejemplo MNC, Mid-size, Startup o Top Tech: ")

# Crear dataframe

custom = pd.DataFrame([{
    "cgpa": cgpa,
    "college_tier": college_tier,
    "python_skill": python_skill,
    "dsa_skill": dsa_skill,
    "ml_skill": ml_skill,
    "web_dev_skill": web_dev_skill,
    "coding_score": coding_score,
    "communication_score": communication_score,
    "aptitude_score": aptitude_score,
    "internships": internships,
    "projects": projects,
    "backlogs": backlogs,
    "resume_score": resume_score,
    "skill_score": skill_score,
    "company_type": company_type,
}])

# One Hot Encoding
custom = pd.get_dummies(
    custom,
    columns=["company_type"],
    dtype=int
)

# Asegurar que tenga las mismas columnas que X_train
custom = custom.reindex(columns=columnas_modelo, fill_value=0)

# Predicción con Random Forest

prediccion_rf = modelo_rf.predict(custom)[0]

print("Predicción con Random Forest")
print(f"Salario predicho: {prediccion_rf:.2f} LPA")


# Predicción con Red Neuronal

prediccion_nn = modelo2.predict(custom).flatten()[0]

print("Predicción con red neuronal")
print(f"Salario predicho: {prediccion_nn:.2f} LPA")
```
